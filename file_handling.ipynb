```python
text="My telephone number is 408-555-1234"
```

```python
phone=re.search(r'\d\d\d-\d\d\d-\d\d\d\d' , text)
```

```python
re.search(r'\d{3}-\d{3}-\d{4}',text)
```

```python
phone_pattern=re.compile(r'(\d{3}) - (\d{3}) - (\d{4})')
```

```python
results=re.search(phone_pattern,text)
```

```python
results.group()
```
