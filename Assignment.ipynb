```python
import requests
from bs4 import BeautifulSoup as bs
import pandas as pd
from textblob import TextBlob

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

url = 'https://www.amazon.com/product-reviews/B08N5WRWNW'  # Example product review URL
page = requests.get(url, headers=headers)

if page.status_code == 200:  # Check if the request was successful
    soup = bs(page.content, 'html.parser')

    # Extract review titles
    title = soup.find_all('a', class_='review-title')

    review_title = [t.get_text(strip=True) for t in title]

    # Extract ratings
    rating = soup.find_all('i', class_='review-rating')
    rate = [r.get_text(strip=True) for r in rating]

    # Extract review body
    review = soup.find_all('span', class_='review-text-content')
    review_body = [r.get_text(strip=True) for r in review]

    # Save data to a DataFrame
    df = pd.DataFrame({
        'Review_Title': review_title,
        'Rate': rate,
        'Review': review_body
    })

    # Perform sentiment analysis on the reviews
    df['Polarity'] = df['Review'].apply(lambda x: TextBlob(str(x)).sentiment.polarity)

    # Save the DataFrame to a CSV file
    df.to_csv("movie_reviews.csv", index=False)

    print(df)
else:
    print(f"Failed to retrieve the page, status code: {page.status_code}")
```

```python
import requests
from bs4 import BeautifulSoup

# URL of the website
url = 'https://www.reddit.com/r/technology/comments/xyz/post_id/'

# Send a GET request to fetch the page content
response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0'})
soup = BeautifulSoup(response.text, 'html.parser')

# Extract specific text content (e.g., comments)
comments = []
for comment in soup.find_all('p'):
    comments.append(comment.get_text())

# Print the extracted comments
for comment in comments:
    print(comment)
```

```python
import praw

# Authenticate with Reddit API
reddit = praw.Reddit(client_id='your_client_id',
                     client_secret='your_client_secret',
                     user_agent='your_user_agent')

# Fetch Reddit comments from a particular thread
submission = reddit.submission(id='post_id')

# Extract comments
comments = []
submission.comments.replace_more(limit=None)
for comment in submission.comments.list():
    comments.append(comment.body)

# Print the comments
for comment in comments:
    print(comment)
```

```python
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

def clean_text(text):
    # Remove special characters and numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # Convert to lowercase
    text = text.lower()
    # Tokenize words
    words = text.split()
    # Remove stopwords
    words = [word for word in words if word not in stopwords.words('english')]
    return ' '.join(words)

# Clean the extracted comments
cleaned_comments = [clean_text(comment) for comment in comments]
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()

for comment in cleaned_comments:
    score = sia.polarity_scores(comment)
    print(f"Comment: {comment}")
    print(f"Sentiment Score: {score}")
```
