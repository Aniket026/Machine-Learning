```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import eig
```

```python
Marks=np.array([[3,4],[2,8],[6,9]])
print(Marks)
```

```python
Marks_df=pd.DataFrame(Marks,columns=['Physics','Math'])
```

```python
plt.scatter(Marks_df["Physics"],Marks_df["Math"])
```

Step 2 :Make data mean centered(Scaling)

```python
MeanByColumn=np.mean(Marks.T,axis=1)
print(MeanByColumn)

Scaled_Data=Marks -MeanByColumn
```

```python
Marks.T
```

```python
Scaled_Data
```

Step =3 find covariance matrix

```python
#find the co-variance matrix
cov_mat=np.cov(Scaled_Data.T)
cov_mat
```

```python
Eval,Evec=eig(cov_mat)
print(Eval)
print(Evec)
```

```python
Projected_data =Evec.T.dot(Scaled_Data.T)
print(Projected_data.T)
```
